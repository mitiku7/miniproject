## 📊 Project Tasks
- **Problem Framing**: Define ride price estimation as a regression problem.
- **Data Exploration**: Inspect dataset, visualize distributions, detect missing values/outliers.
- **Data Cleaning & Feature Engineering**: Encode categorical variables, scale numerical features, handle outliers.
- **Regression Model**: Train and evaluate a Linear Regression model for price prediction.
- **Classification Model**: Create a binary target (high vs low cost) and train Logistic Regression.
- **Model Evaluation**: Compare regression and classification outcomes, identify influential features.
- **Ethical Reflection**: Discuss potential unfair pricing, risks, and dataset limitations.

---

## ⚖️ Reflection
- **Unfair pricing risk**: Surge pricing may disproportionately affect riders in high-demand areas.  
- **Deployment risk**: Biased predictions if training data doesn’t represent all neighborhoods.  
- **Dataset limitation**: Synthetic/small dataset may not capture real-world complexity.
